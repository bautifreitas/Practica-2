## Práctica de NumPy


```python
import matplotlib.pyplot as plt
import numpy as np
```

## Generacion de datos

```python
# Configuración de semilla para que se puedan obtener los mismos resultados luego
np.random.seed(42)

# Generación de datos
N_SAMPLES = 1000

# Distribución Normal (media=100, desvío estándar=50)
MU, SIGMA = 100.0, 50.0
data_normal = np.random.normal(loc=MU, scale=SIGMA, size=N_SAMPLES)

# Distribución Uniforme (rango entre 20 y 80)
LOW, HIGH = 100.0, 150.0
data_uniform = np.random.uniform(low=LOW, high=HIGH, size=N_SAMPLES)
```

## Analisis Estadistico

```python
def calcular_estadisticos(data: np.ndarray) -> dict:
    """Calcula los estadísticos descriptivos principales."""
    return {
        "Media": np.mean(data),
        "Mediana": np.median(data),
        "Varianza": np.var(data),
        "Desvío estándar": np.std(data),
        "Mínimo": np.min(data),
        "Máximo": np.max(data),
    }

stats_normal = calcular_estadisticos(data_normal)
stats_uniform = calcular_estadisticos(data_uniform)

print("=== Estadisticos de la distribucion normal: ===")
for k, v in stats_normal.items():
    print(f"{k:18s}: {v:8.4f}")

print("\n=== Estadisticos de la distribucion uniforme: ===")
for k, v in stats_uniform.items():
    print(f"{k:18s}: {v:8.4f}")
```

## Visualización

```python
# Configuración de la figura
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle(
    "Comparación de Distribuciones Probabilísticas (N = 1,000)",
    fontsize=16,
    fontweight="bold",
    color="black",
)

# Color amarillo/dorado principal
AMARILLO = "#F1C40F"

# Subplot de la distribución normal
axes[0].hist(
    data_normal,
    bins=30,
    color=AMARILLO,
    edgecolor="black",
    linewidth=1.2,
    alpha=0.85,
    density=True,
    label="Muestras",
)
axes[0].axvline(
    stats_normal["Media"],
    color="black",
    linestyle="--",
    linewidth=2,
    label=f"Media ({stats_normal['Media']:.2f})",
)
axes[0].set_title("Distribución Normal (N = 1,000)", fontsize=12, fontweight="bold")
axes[0].set_xlabel("Valor")
axes[0].set_ylabel("Densidad")
axes[0].grid(True, linestyle=":", alpha=0.6, color="black")
axes[0].legend()

# Subplot de la distribución uniforme
axes[1].hist(
    data_uniform,
    bins=30,
    color=AMARILLO,
    edgecolor="black",
    linewidth=1.2,
    alpha=0.85,
    density=True,
    label="Muestras",
)
axes[1].axvline(
    stats_uniform["Media"],
    color="black",
    linestyle="--",
    linewidth=2,
    label=f"Media ({stats_uniform['Media']:.2f})",
)
axes[1].set_title("Distribución Uniforme (N = 1,000)", fontsize=12, fontweight="bold")
axes[1].set_xlabel("Valor")
axes[1].set_ylabel("Densidad")
axes[1].grid(True, linestyle=":", alpha=0.6, color="black")
axes[1].legend()

plt.tight_layout()
plt.show()
```

## Reflexión

Al comparar las distribuciones podemos decir que la distribución normal forma la clásica curva acampanada con una alta concentración de datos alrededor del centro, coincidiendo casi de forma exacta la media y la mediana. Y que la distribución uniforme se distribuye de manera pareja a lo largo de todo el rango, sin preferencia por ningún valor en particular.
